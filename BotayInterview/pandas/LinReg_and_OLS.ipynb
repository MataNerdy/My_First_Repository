```python
import matplotlib.pyplot as plt
import numpy as np

x = np.arange(-6, 6)
a = 2
b = 1

y = a*x + b
plt.plot(x,y,label=f"y={a}x + {b}")
plt.plot([0], [b], "r*", markersize=10)
plt.ylabel("y")
plt.xlabel("x")
plt.ylim(-10, 10)
plt.xlim(-6, 6)
plt.grid()
plt.legend()
```

```python
def draw_ax(a, b, x, ax, ylim=5):
    y = a * x + b
    ax.plot(x, y, label=f'y = {a}x + {b}')
    ax.plot([0], [b], 'r*', markersize=10)

    ax.plot([0, 1], [b, b], 'y', linewidth=4)
    ax.plot([1, 1], [b, b+a], 'y', linewidth=4)

    ax.set_ylabel('y'); ax.set_xlabel('x')
    ax.set_ylim(-ylim, ylim); ax.set_xlim(-5, 5)
    ax.grid()
    ax.legend(prop={'size': 15})
```

```python
fig, ax = plt.subplots(2, 3, figsize=(15, 10))
x = np.arange(-6, 6)

a, b = 0, 0
draw_ax(a, b, x, ax[0][0])
a, b = 1, 0
draw_ax(a, b, x, ax[0][1])
a, b = 0.5, 0
draw_ax(a, b, x, ax[0][2])

a, b = 0, 2
draw_ax(a, b, x, ax[1][0])
a, b = 1, 2
draw_ax(a, b, x, ax[1][1])
a, b = -0.5, -2
draw_ax(a, b, x, ax[1][2])
```

```python
def mserror(X, w1, w0, y):
  y_pred = w1 * X[:, 0] + w0
  return np.sum(y - y_pred) ** 2 / len(y_pred)

def gr_mserror(X, w1, w0, y):
  y_pred = w1 * X[:, 0] + w0
  return np.array([2/len(X)*np.sum((y-y_pred))*(-1),
                   2/len(X)*np.sum((y-y_pred)*(-X[:,0]))])
```

```python
learning_rate = 0.1
w0, w1 = 0, 0
eps = 0.0001

n_w1 = w1
n_w0 = w0

n = 100

for i in range(n):
  c_w0 = n_w0
  c_w1 = n_w1
  n_w0 = c_w0 - learning_rate * gr_mserror(X, c_w1, c_w0, y)[0]
  n_w1 = c_w1 - learning_rate * gr_mserror(X, c_w1, c_w0, y)[1]

  print(f"Iteration: {i}")
  print(f"Current point: {c_w1, c_w0}, Next point: {n_w1, n_w0}")
  print(f"MSE Error: {mserror(X, c_w1, c_w0, y)}")
  print("--------------------------")
  if (abs(n_w1 - c_w1) < eps) & (abs(n_w0 - c_w0) < eps):
    break
```

```python
print(f"Slope True: {coeffs}, Slope trained: {n_w1}")
print(f"Intercept True: {0}, Intercept trained: {n_w0}")
```

```python
X = np.column_stack([X, np.ones((50))])
X
```

```python
def mserror_mat(X, w, y):
  y_pred = X @ w
  return np.sum((y - y_pred)**2)/len(X)

def gr_mserror_mat(X, w, y):
  y_pred = X @ w
  return 2/len(X)*(y - y_pred) @ (-X)
```

```python
weights = np.zeros(X.shape[1])
weights
```

```python
learning_rate = 0.1
eps = 0.0001

n_w = weights

n = 100

for i in range(n):
  c_w = n_w

  n_w = c_w - learning_rate * gr_mserror_mat(X, c_w, y)

  print(f"Iteration: {i}")
  print(f"Current weights: {c_w}, Next weights: {n_w}")
  print(f"MSE Error: {mserror_mat(X, c_w, y)}")
  print("--------------------------")
  if np.linalg.norm(n_w - c_w, ord=2) < eps:
    break
```

```python
print(f"Weights:\nTrue: {coeffs}")
print(f"Trained: {n_w[:-1]}")
print(f"Intercept: True: 0")
print(f"Trained: {n_w[-1]}")
```
