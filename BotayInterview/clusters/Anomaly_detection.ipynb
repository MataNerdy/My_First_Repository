```python
def outliers_3sigma(series):
  mean = series.mean()
  std = series.std()
  lower = mean - 3*std
  upper = mean + 3*std
  return (series < lower) | (series > upper)
```

```python
def outliers_iqr(series):
  q1 = series.quantile(0.25)
  q3 = series.quantile(0.75)
  iqr = q3 - q1
  lower = q1 - 1.5 * iqr
  upper = q3 + 1.5 * iqr
  return (series < lower) | (series > upper)
```

```python
def outlier_std(data, col, threshold=3):
  mean = data[col].mean()
  std = data[col].std()
  upper = mean + threshold * std
  lower = mean - threshold * std
  anomalies = pd.concat([data[col] < lower, data[col] > upper], axis=1).any(axis=1)
  return anomalies, upper, lower
```

```python
def get_columns_outliers(data, columns=None, function=outlier_std, threshold=3):
  if columns is None:
    columns_to_check = data.columns
  else:
    columns_to_check = columns

  outliers = pd.Series(data=[False]*len(data), index=data_features.index, name='is_outlier')
  comparison_table = {}
  for column in columns_to_check:
    anomalies, upper, lower = function(data, column, threshold)
    comparison_table[column] = [upper, lower, sum(anomalies), 100*sum(anomalies)/len(anomalies)]
    outliers[anomalies[anomalies].index] = True

  comparison_table = pd.DataFrame(comparison_table).T
  comparison_table.columns=['upper_bound', 'lower_bound', 'anomalies_count', 'anomalies_percentage']
  comparison_table = comparison_table.sort_values(by='anomalies_percentage', ascending=True)
  return comparison_table, outliers

def anomalies_report(outliers):
  print("Total number of outliers: {}\nPercentage of outliers: {:.2f}%".format(outliers.sum(), 100*outliers.mean()))
```

```python
def outlier_iqr(data, col, threshold=1.5):
  IQR = data[col].quantile(0.75) - data[col].quantile(0.25)
  upper = data[col].quantile(0.75) + threshold * IQR
  lower = data[col].quantile(0.25) - threshold * IQR
  anomalies = pd.concat([data[col] < lower, data[col] > upper], axis=1).any(axis=1)
  return anomalies, upper, lower
```

```python
from sklearn.base import BaseEstimator
from scipy.spatial.distance import cdist
from sklearn.preprocessing import RobustScaler

class DistanceOutliers(BaseEstimator):
  def __init__(self, metric='euclidean', percentile=90):
    self.metric = metric
    self.percentile = percentile

  def fit(self, X):
    self.centroid = np.mean(X, axis=0).values.reshape(-1, 1).T
    distances_train = cdist(X, self.centroid, metric=self.metric).reshape(-1)
    self.threshold = np.percentile(distances_train, self.percentile)

  def predict(self, X):
    distances = cdist(X, self.centroid, metric=self.metric).reshape(-1)
    predictions = (distances > self.threshold).astype(int)
    return predictions
```

```python
scaler = RobustScaler()
scaled_data = pd.DataFrame(
    data=scaler.fit_transform(data_features),
    columns=data_features.columns
)
```

```python
from sklearn.manifold import TSNE

tsne = TSNE(perplexity=50, n_jobs=-1)
X_tsne = tsne.fit_transform(scaled_data)

plt.figure(figsize=(10, 10))
plt.scatter(X_tsne[:, 0], X_tsne[:, 1])
plt.show()
```

```python
euclidian_model = DistanceOutliers(metric='euclidean', percentile=95)
euclidian_model.fit(scaled_data)
euclidian_outliers = euclidian_model.predict(scaled_data)
anomalies_report(euclidian_outliers)
```

```python
plt.figure(figsize=(10, 10))
plt.scatter(X_tsne[:, 0], X_tsne[:, 1], c=euclidian_outliers)
plt.show()
```

```python
cityblock_model = DistanceOutliers(metric='cityblock', percentile=95)
cityblock_model.fit(scaled_data)
cityblock_outliers = cityblock_model.predict(scaled_data)
anomalies_report(cityblock_outliers)

labeled_data = data_features.copy()
labeled_data['is_outlier'] = cityblock_outliers

plt.figure(figsize=(10, 10))
plt.scatter(X_tsne[:, 0], X_tsne[:, 1], c=cityblock_outliers)
plt.show()
```

```python
from sklearn.cluster import DBSCAN

outlier_percentage = 1.

num_clusters = []
anomaly_percentage = []

eps = 0.01
eps_history = [eps]
while outlier_percentage > 0.05:
  model = DBSCAN(eps=eps).fit(scaled_data)
  labels = model.labels_
  num_clusters.append(len(np.unique(labels))-1)
  labels = np.array([1 if label == -1 else 0 for label in labels])
  outlier_percentage = (labels == 1).mean()
  anomaly_percentage.append(outlier_percentage)
  eps += 0.05
  eps_history.append(eps)

model = DBSCAN(eps).fit(scaled_data)
density_outlier = np.array([1 if label == -1 else 0 for label in model.labels_])
```

```python
anomalies_report(density_outlier)
```

```python
iterations = eps_history[:-1]

fig, ax1 = plt.subplots()

color = 'tab:red'
ax1.set_xlabel('epsilon')
ax1.set_ylabel('number of clusters', color=color)
ax1.plot(iterations, num_clusters, color=color)
ax1.tick_params(axis='y', labelcolor=color)

ax2 = ax1.twinx()
color = 'tab:blue'
ax2.set_ylabel('anomaly percentage', color=color)
ax2.plot(iterations, anomaly_percentage, color=color)
ax2.tick_params(axis='y', labelcolor=color)

fig.tight_layout()
plt.show()
```
