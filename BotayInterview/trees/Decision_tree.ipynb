```python
crtr = (df['ap_hi'] <= 120.0)

display(df[crtr], df[~crtr])
```

```python
def entropy(labels):
  labels = list(labels)
  classes = {}
  for l in labels:
    if l not in classes:
      classes[l] = 0
    classes[l] += 1
  ent = 0
  for l in classes:
    p = classes[l] / len(labels)
    ent += p * np.log2(p)

  return -ent

def gain(left_labels, right_labels, root_labels):
  p = float(left_labels.shape[0]) / (left_labels.shape[0] + right_labels.shape[0])
  IG = entropy(root_labels) - p * entropy(left_labels) - (1 - p) * entropy(right_labels)
  return IG
```

```python
IG = gain(df[crtr]['cardio'], df[~crtr]['cardio'], df['cardio'])
IG
```
