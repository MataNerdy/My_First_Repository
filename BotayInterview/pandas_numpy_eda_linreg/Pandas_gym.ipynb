```python
df["New"] = df["Month_to_end_contract"]-df["Lifetime"]
df.loc[df["New"] < 0, "New"] = None
df.dropna(inplace=True)
df
```

### 3.2 Удаляем столбцы

```python
df.drop('Phone', axis=1) # удаляем столбец "Area code"
df
```

### 3.3 Создание столбца с суммой других столбцов

```python
df["Sum of Avgs"] = df["Avg_additional_charges_total"] + df["Avg_class_frequency_total"] # запись в столбец суммы других столбцов
```

```python
df
```

### 3.4 Заполнение столбца данными

```python
df["Sum of Avgs"] = 100 # заполнение столбца 100
df
```

```python
df.loc[df["gender"] == 1, "Sum of Avgs"] = 200
df
```
