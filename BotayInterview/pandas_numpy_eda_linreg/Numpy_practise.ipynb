## Задача 1

#### Напишите программу на NumPy, чтобы проверить, содержится ли каждый элемент одномерного массива в другом массиве.

#### Ожидаемый результат:
####
#### Массив 1: [ 0 10 20 40 60]
####
#### Массив 2: [0, 40]
####
#### Сравнение каждого элемента массива 1 с массивом 2
####
#### Результат: [ True False False True False]

```python
import numpy as np

arr1 = np.array([0, 10, 20, 40, 60])
arr2 = np.array([0, 40])
print(np.isin(arr1, arr2))
```

## Задача 4
#### Напишите программу на NumPy для подсчёта количества вхождений заданного элемента в массив NumPy.
####
#### Пример вывода:
####
#### Исходный массив:
####
#### [10 20 20 20 20 0 20 30 30 30 0 0 20 20 0]
####
#### [4, 1, 7, 3]



```python
arr1 = np.array([10, 20, 20, 20, 20, 0, 20, 30, 30, 30, 0, 0, 20, 20, 0])
arr2 = np.array([0, 10, 20, 30])
arr = []
for i in arr2:
    arr.append(np.count_nonzero(arr1 == i))
print(arr)
```

## Задача 9
#### Напишите программу на NumPy для вычисления разности между соседними элементами массива поэлементно, а также добавления [0, 0] в начало и [200] в конец данного массива.
####
#### Пример вывода:
####
#### Исходный массив:
####
#### [1 3 5 7 0]
####
#### Разность между соседними элементами массива поэлементно, добавив [0, 0] в начало и [200] в конец массива:
####
#### [ 0 0 2 2 2 -7 200]
####
#### Примечание: разность между соседними элементами означает: $out[i] = a[i+1] - a[i]$

```python
arr1 = np.array([1, 3, 5, 7, 0])
arr = np.diff(arr1)
arr = np.insert(np.append(arr, 200), 0, [0, 0])
print(arr)
```

## Задача 11
#### Вычислите евклидово расстояние между двумя векторами a и b.
####
#### Ввод:
####
#### [1,2,3,4,5]
####
#### [4,5,6,7,8]
####

```python
### Open Your CSV file and print the first 10 rows in a good format ###
import pandas as pd
df = pd.read_csv("/content/tmdb_5000_movies.csv")
df
```

```python
### Check if any NANs in your dataset and fill them with a good filer ###
df.isna().sum()
```

```python
#drop homepage (unnecessary column)
df = df.drop(columns=['homepage'])
#impute null values in tagline column with unknown
df['tagline'] = df['tagline'].fillna("unknown")
#drop the rest null values
df = df.dropna()
df
```

```python
print("After imputation:\n",df.isna().sum())
```

```python
### Discover the types of each columns and modify them if needed ###
print(df.dtypes)
df['release_date'] = pd.to_datetime(df['release_date'])
print(df.info())
```

```python
### Give a very simple statistical analsys for the numerical columns ###
df.describe().T
```

```python
#create a release_year from the date values in release_date column
df['release_year'] = pd.DatetimeIndex(df['release_date']).year
```

```python
df_silce = df.groupby(['release_year'], as_index=False).vote_average.mean()
### Calculate the mean rate for [1999, 1980, 2004, 2017] the movies grouped by years ###
df_silce[df_silce['release_year'].isin([1999, 1980, 2004, 2017])]
```

```python
### Ex: Average Movies Rate at 2009 is 6.065587 ###
df_silce.loc[df_silce['release_year']==2009, 'vote_average'].values[0].round(6)
```

```python
### Rearange the dataframe based on revenue - budget values ###
df["Profit"] = df['revenue'] - df['budget']
df_sorted = df.sort_values(["Profit"], ascending=False)
df_sorted.head()
```

```python
import ast
### Find the top 2 countries with the highest production movies number ###
df['countries'] = df['production_countries'].apply(lambda x: [i["name"] for i in ast.literal_eval(x)])
df['countries'] = [','.join(map(str, l)) for l in df['countries']]
dfc = df['countries'].value_counts().sort_values(ascending=False).head(2)

#convert list item to string
print(str(dfc.index.values))
```

```python
import ast
### Find the top 1 company with the highest production movies number ###
df['companies'] = df['production_companies'].apply(lambda x: [i["name"] for i in ast.literal_eval(x)])
df['companies'] = [','.join(map(str, l)) for l in df['companies']]
dfc = df['companies'].value_counts().sort_values(ascending=False).head(10)
print(str(dfc[1:3].index.values))
```

```python
### Is there any relation between the runtime and average vote value ### **
#scatter plot is useful for examining the relationship between two numerical variables.

import matplotlib.pyplot as plt
y = df['runtime']
x = df['vote_average']
#add points to plot
plt.scatter(x, y, c="green", alpha=0.5)
#find line of best fit
a, b = np.polyfit(x, y, 1)
#add line of best fit to plot
plt.plot(x, a*x+b, c="b", linestyle="--")
#add fitted regression equation to plot
plt.text(1, 50, f'y = {a:.2f} * x + {b:.2f}')
plt.xlabel("Vote Average")
plt.ylabel("Runtime")
plt.show()
```

```python
corr = x.corr(y)
corr
```
